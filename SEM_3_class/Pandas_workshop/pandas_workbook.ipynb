### Combining Conditions
Real questions are often more complicated.

For example:

Find students who scored more than 80 in both Maths and Python.

In normal Python, we might think about using:
> and

But Pandas works differently when comparing an entire Series.

For element-by-element logical operations, we use:

| Operation | Pandas operator |   |
| --------- | --------------- | - |
| AND       | `&`             |   |
| OR        | l           |   |
| NOT       | `~`             |   |


When combining conditions, **put each condition inside parentheses.**

```python 
df[(df['Physics']>80) & (df['Maths']>84)]
df[(df['Physics']>80) | (df['Maths']>84)]
df[~(df['Physics']>80)]

### Handling Missing Data: dropna()

`dropna()` removes rows or columns containing missing values.

**Strategies:**
- Drop entire rows with ANY missing value
- Drop entire columns with ANY missing value
- Drop rows where ALL values are missing
- Drop columns where ALL values are missing

**Syntax:**
```python
df.dropna()                           # Remove rows with any NaN
df.dropna(axis=1)                     # Remove columns with any NaN
df.dropna(how='all')                  # Remove rows where ALL values are NaN
df.dropna(subset=['column_name'])     # Remove rows where specific column has NaN
df.dropna(thresh=3)                   # Keep rows with at least 3 non-null values
```

**Warning:** Dropping rows/columns removes data! Use with caution.

### Handling Missing Data: fillna()

`fillna()` replaces missing values with specified values instead of removing data.

This is often better than dropping data because it preserves row/column count.

**Methods:**
```python
df.fillna(0)                          # Replace all NaN with 0
df.fillna('Unknown')                  # Replace with a specific value
df.fillna(method='ffill')             # Forward fill: use previous value
df.fillna(method='bfill')             # Backward fill: use next value
df['column'].fillna(df['column'].mean())  # Fill with mean/median/mode
df.fillna({'col1': 0, 'col2': 'N/A'}) # Fill different columns differently
```

**Forward Fill (ffill):** Propagates the last known value forward
**Backward Fill (bfill):** Propagates the next known value backward

### Grouping Data with groupby()

`groupby()` splits data into groups based on one or more columns, then applies operations to each group.

This is powerful for answering questions like:
- "What is the average score per department?"
- "How many students are in each department?"
- "What is the total sales per region?"

**Basic Syntax:**
```python
df.groupby('column').operation()
```

**Common Operations:**
```python
df.groupby('Department').mean()       # Average for each group
df.groupby('Department').sum()        # Sum for each group
df.groupby('Department').count()      # Count of rows in each group
df.groupby('Department').min()        # Minimum value in each group
df.groupby('Department').max()        # Maximum value in each group
df.groupby('Department').agg(['mean', 'sum', 'count'])  # Multiple operations
```

**Grouping by Multiple Columns:**
```python
df.groupby(['Department', 'Gender']).mean()  # Nested grouping
```
