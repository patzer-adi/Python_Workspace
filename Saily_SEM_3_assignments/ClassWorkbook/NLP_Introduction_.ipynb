## Section 2: Basic Text and Sample Data

```python
# Create sample text
sample_text = """Natural Language Processing (NLP) is fascinating! It's used everywhere today.
From chatbots to translation services, NLP powers modern AI applications.
Processing human    language is complex but rewarding.
for more details,   visit https://en.wikipedia.org/wiki/Natural_language_processing.
contact me at john.doe@example.com"""

print("Original Text:")
print(sample_text)
print("\n" + "="*60)
```

## Section 3: Text Cleaning - Step by Step

```python
# Step 1: Convert to lowercase
text_lower = sample_text.lower()
print(text_lower)
```

```python
# Step 2: Remove URLs
text_no_urls = re.sub(r'http\S+|www\S+|https\S+', '', text_lower)
print(text_no_urls)
```

```python
# Step 3: Remove email addresses
text_no_emails = re.sub(r'\S+@\S+', '', text_no_urls)
print(text_no_emails)
```

```python
# Step 4: Remove special characters and numbers (keep only letters and spaces)
text_no_special = re.sub(r'[^a-zA-Z\s]', '', text_no_emails)
print(text_no_special)
```

```python
# Step 5: Remove extra whitespace
text_cleaned = re.sub(r'\s+', ' ', text_no_special).strip()
print(text_cleaned)
```

## Section 4: Tokenization

```python
# Sentence Tokenization
sentences = sent_tokenize(sample_text)
```

```python
sentences
```

```python
# Word Tokenization on cleaned text
words = word_tokenize(text_cleaned)
```

```python
words
```

## Section 5: Stop Words Removal

```python
# Get English stop words
stop_words_set = set(stopwords.words('english'))
stop_words_set
```

```python
# Check which words in our text are stop words
stop_words_in_text = [word for word in words if word in stop_words_set]
stop_words_in_text
```

```python
# Remove stop words
words_filtered = [word for word in words if word not in stop_words_set]
words_filtered
```

## Section 6: Stemming

```python
# Initialize Porter Stemmer
stemmer = PorterStemmer()

# Example words to demonstrate stemming
example_words = ['running', 'runs', 'ran', 'processing', 'processed', 'processes']

for word in example_words:
    stemmed = stemmer.stem(word)
    print(stemmed)
```

```python
# Apply stemming to our filtered tokens
stemmed_words = []

for word in words_filtered:
    stemmed_word = stemmer.stem(word)
    stemmed_words.append(stemmed_word)

print(stemmed_words)
```

## Section 7: Lemmatization

```python
# Initialize WordNet Lemmatizer
lemmatizer = WordNetLemmatizer()

# Example words for lemmatization
example_words_lemma = ['running', 'runs', 'is', 'was', 'better', 'good', 'processing', 'processed']

for word in example_words_lemma:
    lemmatized = lemmatizer.lemmatize(word, pos='v')
    print(lemmatized)
```

```python
# Apply lemmatization to our filtered tokens
lemmatized_words = []

for word in words_filtered:
    lemmatized_word = lemmatizer.lemmatize(word, pos='v')
    lemmatized_words.append(lemmatized_word)

print("\nLemmatized Filtered Tokens:")
print(lemmatized_words)
```

## Section 8: Stemming vs Lemmatization Comparison

```python
# Create comparison DataFrame
comparison_words = ['running', 'runs', 'is', 'was', 'processing', 'processed', 'better', 'good']

comparison_data = []

for word in comparison_words:
    stemmed = stemmer.stem(word)
    lemmatized = lemmatizer.lemmatize(word, pos='v')
    comparison_data.append([word, stemmed, lemmatized])

comparison_df = pd.DataFrame(comparison_data, columns=['Original', 'Stemmed', 'Lemmatized'])

print("\nStemming vs Lemmatization Comparison:")
print(comparison_df.to_string(index=False))
```
