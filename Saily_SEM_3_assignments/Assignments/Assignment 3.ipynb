```python
#Problem 2. Temperature Analysis

arr = np.array([32.5, 31.8, 30.2, 33.1, 34.5, 29.8, 31.0])

#print(arr)
sum = 0
#
for i in arr:
    sum += i
#print (sum)
def avg(arr):
    average = sum/arr.size
    return average

print("Average:",avg(arr))   
print("Maximum Temperature:", np.max(arr))
print("Minimum Temperature:", np.min(arr))
print("Day with maximum temperature:", np.argmax(arr))

average = avg(arr)
for i in arr:
    if i > average:        
        print("Temperature greater then average:",i)
```

```python
#Problem 5. Numbers Above a Threshold

genarr = np.random.randint(1,101, size = 30)
arr50 = []
arr2550 = []
add = 0
cnt = 0
print("Array:", genarr)

for i in genarr:
    if i > 50:
        arr50.append(int(i))
        add += i
        cnt += 1
    elif i>25 & i<50:
        arr2550.append(int(i))
   
print("Numbers greater than 50:",arr50)
print("Numbers greater than 25 and smller than 50:",arr2550)
print("Count of numbers greater than 50:",cnt)
print("Average:", add/cnt)
```

```python
#Problem 6. Student Performance Matrix

marks = np.array([
    [78, 82, 91],   # Student 1
    [65, 74, 80],   # Student 2
    [92, 88, 95],   # Student 3
    [56, 69, 72],   # Student 4
    [84, 90, 87]    # Student 5
])

student_avg = np.mean(marks, axis=1)
print("Average marks of each student:", student_avg)

subject_avg = np.mean(marks, axis=0)
print("Average marks of each subject:", subject_avg)

# Highest mark in entire dataset
print("Highest mark:", np.max(marks))
print("Lowest mark:", np.min(marks))
highest_student = np.argmax(student_avg)
print("Student with highest average: Student", highest_student + 1)
highest_subject = np.argmax(subject_avg)
subjects = ["Mathematics", "Physics", "Programming"]
print("Subject with highest average:", subjects[highest_subject])
```

```python
#Problem 8. Scientific Data Transformation

import numpy as np

def calculate(arr):
    square = arr ** 2
    square_root = np.sqrt(arr)
    exponential = np.exp(arr)

    print("Square:", square)
    print("Square root:", square_root)
    print("Exponential:", exponential)

    print("Mean of square:", np.mean(square))
    print("Mean of square root:", np.mean(square_root))
    print("Mean of exponential:", np.mean(exponential))


arr = np.array([2, 4, 6, 8, 10, 12])

calculate(arr)
```

```python
#Problem 13. Experiment Comparison

experiment_A = np.array([12, 15, 14, 18, 20, 17, 16])
experiment_B = np.array([10, 13, 15, 16, 19, 21, 18])
print("Mean of A:", np.mean(experiment_A))
print("Mean of B:", np.mean(experiment_B))

print("Standard deviation of A:", np.std(experiment_A))
print("Standard deviation of B:", np.std(experiment_B))

print("Maximum of A:", np.max(experiment_A))
print("Maximum of B:", np.max(experiment_B))

print("Minimum of A:", np.min(experiment_A))
print("Minimum of B:", np.min(experiment_B))

difference = experiment_A - experiment_B
print("Difference:", difference)

if np.std(experiment_A) > np.std(experiment_B):
    print("Experiment A shows greater variation")
else:
    print("Experiment B shows greater variation")
```

```python
#Problem 14. Mini Data Analysis Challenge

def create_data():
    data = np.random.randint(1, 100, size=(10, 5))
    return data

def analyze_data(data):

    print("Dataset:")
    print(data)

    print("\nShape:", data.shape)
    print("Dimensions:", data.ndim)

    print("\nMean of each variable:")
    print(np.mean(data, axis=0))

    print("\nMean of each experiment:")
    print(np.mean(data, axis=1))

    print("\nMaximum:", np.max(data))
    print("Minimum:", np.min(data))

    overall_mean = np.mean(data)
    print("\nOverall mean:", overall_mean)
    print("Values above overall mean:")
    print(data[data > overall_mean])

    reshaped = data.reshape(5, 10)
    print("\nReshaped data (5 × 10):")
    print(reshaped)
    print("New shape:", reshaped.shape)

    flattened = reshaped.flatten()
    print("\nFlattened data:")
    print(flattened)
    print("Flattened shape:", flattened.shape)

data = create_data()
analyze_data(data)
```

```python
#Problem 15. Final Challenge: Student Performance Analysis

def generate_marks():
    return np.random.randint(0, 101, size=(10, 4))

def calculate_student_average(marks):
    return np.mean(marks, axis=1)

def calculate_subject_average(marks):
    return np.mean(marks, axis=0)

def get_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

marks = generate_marks()

student_avg = calculate_student_average(marks)
subject_avg = calculate_subject_average(marks)

highest_student = np.argmax(student_avg)
lowest_student = np.argmin(student_avg)

highest_mark = np.max(marks)
lowest_mark = np.min(marks)
students_above_75 = np.sum(student_avg > 75)
marks_above_90 = marks[marks > 90]

students = []

for i in range(10):
    student = {
        "Student": i + 1,
        "Marks": marks[i],
        "Average": student_avg[i],
        "Grade": get_grade(student_avg[i])
    }

    students.append(student)

print("STUDENT PERFORMANCE ANALYSIS")

print("\nMarks:")
print(marks)

print("\nAverage mark of each student:")
for i in range(10):
    print(f"Student {i + 1}: {student_avg[i]:.2f}")

print("\nAverage mark of each subject:")
for i in range(4):
    print(f"Subject {i + 1}: {subject_avg[i]:.2f}")

print("\nStudent with highest average:")
print(f"Student {highest_student + 1} ({student_avg[highest_student]:.2f})")

print("\nStudent with lowest average:")
print(f"Student {lowest_student + 1} ({student_avg[lowest_student]:.2f})")

print("\nHighest mark in dataset:", highest_mark)
print("Lowest mark in dataset:", lowest_mark)

print("\nNumber of students with average above 75:",
      students_above_75)

print("\nAll marks greater than 90:")
print(marks_above_90)

print("\nStudent Grades:")
for student in students:
    print(
        f"Student {student['Student']}: "
        f"Average = {student['Average']:.2f}, "
        f"Grade = {student['Grade']}"
    )
```
