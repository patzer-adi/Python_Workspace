```python
# Problem 8 Bank Account Methods

class BankAccount:
    def __init__(self, account_num, owner, balance):
        self.account_num = account_num
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        print("Deposited:", amount)
        print("New Balance:", self.balance)


    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount
            print("Withdrawn:", amount)
            print("New Balance:", self.balance)
        else:
            print("Insufficient balance")

account1 = BankAccount(101, "Siya", 70000)
account1.deposit(10000)
account1.withdraw(20000)
account1.withdraw(100)       
```

```python
# Problem 9. Withdrawal Status

class BankAccount:
    def __init__(self, account_num, owner, balance):
        self.account_num = account_num
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount
            return True
        else:
            return False

account1 = BankAccount(101, "Siya", 70000)
status1 = account1.withdraw(20000)

print("Valid withdrawal status:", status1)
print("Account Balance:", account1.balance)
status2 = account1.withdraw(60000)

print("Invalid withdrawal status:", status2)
print("Account Balance:", account1.balance)
```

```python
# Problem 10. Display Account Details

class BankAccount:
    def __init__(self, account_num, owner, balance):
        self.account_num = account_num
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount
            return True
        else:
            return False

    def __str__(self):
        return f"Account Number: {self.account_num}\nOwner: {self.owner}\nBalance: {self.balance}"

account = BankAccount(101, "Siya", 70000)
print(account)
```

```python
#Problem 11. Student Class

class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def average_marks(self):
        return sum(self.marks) / len(self.marks)

student1 = Student("Siya", 101, [80, 75, 90])
student2 = Student("Diya", 102, [85, 90, 95])
student3 = Student("Riya", 103, [70, 80, 75])

print("Student 1:", student1.name)
print("Average Marks:", student1.average_marks())

print("Student 2:", student2.name)
print("Average Marks:", student2.average_marks())

print("Student 3:", student3.name)
print("Average Marks:", student3.average_marks())
```

```python
# Problem 13. Method Overriding

class Vehicle:
    def move(self):
        print("Vehicle is moving")


class Car(Vehicle):
    def move(self):
        print("Car is driving")


class Boat(Vehicle):
    def move(self):
        print("Boat is sailing")

vehicle = Vehicle()
car = Car()
boat = Boat()
vehicle.move()
car.move()
boat.move()
```

```python
# Problem 14. Using super()

class Vehicle:
    def __init__(self, brand):
        self.brand = brand


class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors

car1 = Car("Toyota", 4)
car2 = Car("BMW", 2)
print("Car 1")
print("Brand:", car1.brand)
print("Doors:", car1.doors)
print("Car 2")
print("Brand:", car2.brand)
print("Doors:", car2.doors)
```
