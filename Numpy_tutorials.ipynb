```python
mat2d = np.arange(25).reshape(5,5)
mat2d
```

```python
flaten_2d = mat2d.flatten()
flaten_2d
```

```python
ravel_2d = mat2d.ravel()
ravel_2d
```

```python
print(flaten_2d)
```

```python

# Modifying a Ravel array updates the parent matrix
ravel_2d[0] = 99
print(mat2d[0, 0]) 
print(mat2d)
print(ravel_2d) 


# Output: 99 (The original matrix changed!)

# Modifying a Flatten array protects the parent matrix
flaten_2d[1] = 88
print(mat2d[0, 1]) 
# Output: 2 (The original matrix is unchanged)
print(mat2d)
print(flaten_2d) 
```
