```python
# problem 3

def temp_analysis(incoming_array):
    temp_array = np.array(incoming_array)
    days = ['Sun', 'Mon', 'Tues', 'Wed', 'Thrus', 'Fir', 'Sat']
    print(f"Average temperature: {np.average(temp_array)}")
    print(f"Maximum Temperature: {np.max(temp_array)}")
    print(f"Minimum temperatur: {np.min(temp_array)}")
    print(f"Day with max temperature:  Day {days[np.argmax(temp_array)]}") 
    # argmax return the index of the element
    print(f"Days with more than average temp: ")
    avg = np.average(temp_array)
    for i in range(0,temp_array.size):
        if temp_array[i]  > avg:
            print(f"{temp_array[i]}")
    
    
```

```python
temp_list = [32.5, 31.8, 30.2, 33.1, 34.5, 29.8, 31.0]
temp_analysis(temp_list)
```

```python
marks = np.array([
    [78, 82, 91],
    [65, 74, 80],
    [92, 88, 95],
    [56, 69, 72],
    [84, 90, 87]
])

subjects = ["Mathematics", "Physics", "Programming"]
students = ["Student 1", "Student 2", "Student 3", "Student 4", "Student 5"]

df = pd.DataFrame(
    marks,
    columns=subjects,
)

df["Student"] = students
df.set_index("Student", inplace=True)

print("Student marks:")
print(df)

print("Average marks of each student:")
student_avg = []

for student in df.index:
    avg = df.loc[student].mean()
    student_avg.append(avg)
    print(f"{student}: {avg}")
```

```python

print("Average marks in each subject:")
subject_avg = []

for subject in df.columns:
    avg = df[subject].mean()
    subject_avg.append(avg)
    print(f"{subject}: {avg}")
```

```python

print(f"Highest mark: {df.max().max()}")
print(f"Lowest mark: {df.min().min()}")

highest_student = df.mean(axis=1).idxmax()
print(f"Student with highest average: {highest_student}")

highest_subject = df.mean(axis=0).idxmax()
print(f"Subject with highest average: {highest_subject}")
```

```python
# problem 8

def data_transformation(incoming_array):

    data = np.array(incoming_array)

    squared = np.square(data)
    square_root = np.sqrt(data)
    exponential = np.exp(data)

    print(f"Original data: {data}")

    print(f"\nSquare of measurements: {squared}")
    print(f"Mean of squared data: {np.mean(squared)}")

    print(f"\nSquare root of measurements: {square_root}")
    print(f"Mean of square root data: {np.mean(square_root)}")

    print(f"\nExponential of measurements: {exponential}")
    print(f"Mean of exponential data: {np.mean(exponential)}")
```

```python

measurements = [2, 4, 6, 8, 10, 12]

data_transformation(measurements)
```

```python
# problem 10
experiment_1 = np.array([10, 20, 30, 40, 50])
experiment_2 = np.array([15, 25, 35, 45, 55])

print("Experiment 1:")
print(experiment_1)

print("\nExperiment 2:")
print(experiment_2)
```

```python
# Single 1-D array
combined_1d = np.concatenate((experiment_1, experiment_2))

print("\nCombined 1-D array:")
print(combined_1d)
print(f"Shape: {combined_1d.shape}")
```

```python
# Experiments as separate rows
combined_rows = np.vstack((experiment_1, experiment_2))

print("\nExperiments as separate rows:")
print(combined_rows)
print(f"Shape: {combined_rows.shape}")
```

```python
# Experiments as separate columns
combined_columns = np.column_stack((experiment_1, experiment_2))

print("\nExperiments as separate columns:")
print(combined_columns)
print(f"Shape: {combined_columns.shape}")
```

```python
# problem 14

def analyze_data(data):

    print("Dataset:")
    print(data)

    print(f"\nShape: {data.shape}")
    print(f"Dimensions: {data.ndim}")

    print("\nMean of each variable:")
    for i in range(data.shape[1]):
        print(np.mean(data[:, i]))

    print("\nMean of each experiment:")
    for i in range(data.shape[0]):
        print(np.mean(data[i]))

    print(f"\nMaximum value: {np.max(data)}")
    print(f"Minimum value: {np.min(data)}")

    overall_mean = np.mean(data)
    print(f"\nOverall mean: {overall_mean}")
    print(f"Values above overall mean: {data[data > overall_mean]}")
```

```python
def reshape_data(data):

    reshaped = data.reshape(5, 10)

    print("\nReshaped data:")
    print(reshaped)

    flattened = reshaped.flatten()

    print("\nFlattened data:")
    print(flattened)
```

```python
data = np.random.randint(1, 101, (10, 5))

analyze_data(data)
reshape_data(data)
```
