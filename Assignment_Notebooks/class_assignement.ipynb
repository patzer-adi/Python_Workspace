```python
#problem 1

marks_list = []

for i in range(0,5):
    sub_name = input(f"Enter subject {i+1} name")
    sub_mark = int(input(f"Enter subject {i+1} marks"))
    marks_list.append(sub_mark)

average = sum(marks_list) / len(marks_list)
print(average)

if average >= 80: print("A Grade")
elif average < 80 and average >= 60: print("B Grade")
elif average < 60 and average >= 40: print("C Grade")
else:
    print("F Grade")
```

```python
#problem 2

string_sentence = input("Enter a sentence")
word_count = len(string_sentence.split(' '))
print(f"This is the word count i have: {word_count}")

char_count = len(string_sentence)
print(f"This is the char count i have: {char_count}")

vowel = ['a', 'e', 'i', 'o', 'u']
vowel_count = 0
consonant_count = 0
for ch in string_sentence:
    if ch in vowel:
        vowel_count+=1
    else:
        consonant_count += 1

print(f"This is the vowel count i have: {vowel_count}")
print(f"This is the consonant count i have: {consonant_count}")
```

```python
# problem 3

user_list_intger = []

end = int(input("Enter the length of a list"))

for i in range(0, end):
    num = int(input("Enter the number:"))
    user_list_intger.append(num)

print(user_list_intger)

even_list = []
odd_list = []

for i in user_list_intger:
    if i % 2 == 0:
        even_list.append(i)
    else:
        odd_list.append(i)

print(f"Even list: {even_list}")
print(f"Odd list : {odd_list}")
```

```python
#problem 4

user_list_integer = []

end = int(input("Enter the length of a list"))

for i in range(0, end):
    num = int(input("Enter the number:"))
    user_list_integer.append(num)

print(user_list_intger)

numbers = user_list_integer

total = 0

largest = numbers[0]
smallest = numbers[0]

for number in numbers:
    total = total + number

    if number > largest:
        largest = number
    if number < smallest:
        smallest = number
average = total / len(numbers)

print(f"Largest: {largest}")
print(f"Smallest: {smallest}")
print(f"Total: {total}")
print(f"Average: {average}")
```

```python
#problem 6

mark_len = int(input("Enter the length of list"))

marks_list = [int(input(f"Enter marks {i+1}")) for i in range(mark_len)]

marks_tuple = tuple(marks_list)

print(marks_tuple)

average = sum(marks_tuple) / len(marks_tuple)
max_marks = max(marks_tuple)
lowest_marks = min(marks_tuple)

print(f"Maximum Marks {lowest_marks}")
print(f"Lowest Marks {max_marks}")
count = 0
for i in marks_tuple:
    if i > average:
        count += 1
print(f"Number of students scoring above average: {count}")
```

```python
# problem 9

set1 = set([input("Enter set 1:") for i in range(0,5)])
set2 = set(input("Enter set 2:")  for i in range(0,5) )
print(f"Union:{set1.union(set2)}") # s1 | s2
print(f"Intersection: {set1.intersection(set2)}") # s1 & s2
print(f"Difference: {set1.difference(set2)}") #s1 - s2
print(f"Symmetric difference {set1.difference(set2)}") #s1 ^ s2
```

```python
# # problem 10
# contact book

contacts = {}

while True:
    print("------ Contact Book ------")
    print(" '1': Add, '2': Search, '3': Update, '4':Delete, '5':Display, '6': Exit ")
    choice = input("Enter your choice")[0]
    if choice == '1':
        print("Add a contact")
        name = input("Enter name of contact")
        phone_no = int(input("Enter contact number"))
        contacts.update({name:phone_no})
    elif choice == '2':
        name = input("Enter name to search:")
        if name in contacts:
            print(f"Phone: {contacts[name]}")
        else:
            print("Contact not found")
    elif choice == '3':
        name = input("Enter name to update:")
        if name in contacts:
            number = input("Enter new phone number:")
            contacts[name] = number
        else:
            print("Contact not found.")
    elif choice == "4":
        name = input("Enter name to delete:")
        if name in contacts:
            del contacts[name]
            print("Contact deleted sucessfully")
        else:
            print("Contact not found")
    elif choice == "5":
        if len(contacts) == 0:
            print("No contacts available")
        else:
            for name, number in contacts.items():
                print(name,":",number)
    elif choice == "6":
        print("Exiting Contact Book...")
        break
    else:
        print("Invalid choice")
            
```

```python
# problem 14

len_list = int(input("Length of a list"))

original_list = [int(input("Enter list elements (okay with duplicates):")) for i in range(len_list) ]

unique_list = []

for num in original_list:
    if num not in unique_list:
        unique_list.append(num)

print(f"Original list:{original_list}")
print(f"Unique list: {unique_list}")
```

```python
# Problem 15

students = {}

while True:
    print("\n--- Student Record Management System ---")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display all students")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        roll_no = int(input("Enter roll number: "))
        name = input("Enter student name: ")
        course = input("Enter course name: ")
        marks = int(input("Enter marks: "))

        students.update({
            roll_no: {
                "Name": name,
                "Course": course,
                "Marks": marks
            }
        })

        print("Student added successfully")

    elif choice == "2":
        roll_no = int(input("Enter roll number to search: "))

        if roll_no in students:
            print(f"Name: {students[roll_no]['Name']}")
            print(f"Course: {students[roll_no]['Course']}")
            print(f"Marks: {students[roll_no]['Marks']}")
        else:
            print("Student not found")

    elif choice == "3":
        roll_no = int(input("Enter roll number to update: "))

        if roll_no in students:
            name = input("Enter new name: ")
            course = input("Enter new course: ")
            marks = int(input("Enter new marks: "))

            students[roll_no]["Name"] = name
            students[roll_no]["Course"] = course
            students[roll_no]["Marks"] = marks

            print("Student updated successfully")

        else:
            print("Student not found")

    elif choice == "4":
        roll_no = int(input("Enter roll number to delete: "))

        if roll_no in students:
            del students[roll_no]
            print("Student deleted successfully")
        else:
            print("Student not found")

    elif choice == "5":
        if len(students) == 0:
            print("No student records available")

        else:
            for roll_no in students:
                print("\nRoll Number:", roll_no)
                print("Name:", students[roll_no]["Name"])
                print("Course:", students[roll_no]["Course"])
                print("Marks:", students[roll_no]["Marks"])

    elif choice == "6":
        print("Exiting...")
        break

    else:
        print("Invalid choice")
```
