```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("allenai/specter2_base")
print("SPECTER2 loaded successfully!")
```

```python
text = "A machine learning system for detecting plant diseases from images."

embedding = model.encode(text)

print(type(embedding))
print(embedding.shape)
```

```python
texts = [
    "A wheel-based mechanism for transporting objects.",
    "A vehicle using wheels to move goods from one location to another.",
    "A method for growing plants using controlled indoor lighting.",
]

embeddings = model.encode(
    texts,
    normalize_embeddings=True
)

print("Shape:", embeddings.shape)
print("Embedding type:", type(embeddings))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(embeddings)

print(similarity)
```
