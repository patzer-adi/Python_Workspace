## Experiment 2 - String Indexing and Slicing

```python
food = "I like Dum Biryani"          # create a string

print(food[0])                        # indexing: first character (index 0) 'I'
print(food[-1])                       # indexing: last character (index -1) 'i'
print(food[7:11])                     # slicing: characters from index 7 to 10 'Dum '
print(food[::-1])                     # slicing with step -1: reverses the whole string
```

## Experiment 3 - List append() and remove()

```python
numbers = [5, 15, 25]                # create a list

numbers.append(35)                    # append(): adds 35 to the end  
print(numbers)

numbers.remove(15)                    # remove(): deletes the first occurrence of 15  
print(numbers)

numbers.insert(1, 100)                # insert(): places 100 at index 1  
print(numbers)

numbers.sort()                        # sort(): arranges the list in ascending order
print(numbers)
```

## Experiment 4 - More String Slicing

```python
text = "Kalyan"                       # create a string

print(text[0:5])                      # slicing: characters from index 0 to 4  
print(text[3:])                       # slicing: from index 3 to the end -> 'yan'
print(text[:6])                       # slicing: from start up to (not including) index 6 
print(text[::-1])                     # slicing with step -1: reverses the whole string
```

## Experiment 7 - Lists are Mutable

```python
numbers = [10, 20, 30]
numbers[1] = 200
print(numbers)
# Why can a list be modified while a string cannot?
```

Lists can be changed after creation (mutable). 

Strings cannot any change creates a new string instead (immutable).

## Experiment 9 - Student ID Card

```python
name = input("Enter Name: ")            # take name as input (string)
age = input("Enter Age: ")              # take age as input (string)
roll_no = input("Enter Roll Number: ")  # take roll number as input (string)
course = input("Enter Course: ")        # take course as input (string)

print("=" * 28)              # top border of the card
print("STUDENT CARD")        # card title
print("=" * 28)              # separator line

print("Name       :", name)      # display name
print("Age        :", age)       # display age
print("Roll No.   :", roll_no)   # display roll number
print("Course     :", course)    # display course
print("=" * 28)                  # bottom border of the card

print()                           # blank line for spacing

print(f"Hello {name}!")                    # f-string: inserts name directly into the message
print("Welcome to the world of Python.")

print('"Python is easy to learn!"')        # double quotes shown using single-quoted string
print("'I\'ll practice every day.'")       # backslash escapes the apostrophe inside single quotes
print('"Practice makes perfect."')         # double quotes shown using single-quoted string

print("\\/" * 12)             # backslash-escaped '\/' repeated 12 times as a decorative line
print("Happy Coding!")        # closing message
```

```python
# 1. Single quotes - simplest case
name = 'Alice'

# 2. Double quotes - identical to single quotes
name2 = "Alice"
print(name == name2)   # True : no difference

# 3. Single quotes containing a double quote no escaping needed
line1 = 'She said "hello" to me'

# 4. Double quotes containing an apostrophe no escaping needed
line2 = "It's a sunny day"

# 5. Single quotes containing an apostrophe : needs a backslash escape
line3 = 'It\'s a sunny day'

# 6. Triple quotes — can contain BOTH ' and " freely so no escaping
line4 = """She said "it's fine" and walked away."""

# 7. Triple quotes the real superpower is : multi-line strings
poem = """Roses are red,
Violets are blue,
No escaping needed,
Even with "quotes" or it's true."""

print(line1)
print(line2)
print(line3)
print(line4)
print(poem)
```

```python
# --- String Creation ---
s1 = 'Hello'
s2 = "World"
s3 = '''This is a
multi-line string'''
print(s1, s2)
print(s3)

# --- Indexing ---
text = "Python"
print(text[0])    # first character
print(text[-1])   # last character

# --- Slicing ---
print(text[0:4])   # characters 0 to 3
print(text[::-1])  # reversed string

# --- String Operations ---
print(s1 + " " + s2)        # concatenation
print(s1 * 3)                # repetition
print("Py" in text)          # containment check
print(len(text))             # length
print(text.upper())          # uppercase
print(text.lower())          # lowercase
print("  padded  ".strip())  # strip whitespace
print("a,b,c".split(","))    # split
```

```python
colors = ["Red", "Green", "Blue"]
print(colors)

student = ["Adi", 21, 8.5, True]   # string, int, float, boolean together mixed data types
print(student)

print(colors[0])    # Indexing : first item
print(colors[-1])   # last item

print(colors[0:2])  # silicing items at index 0 and 1

scores = [45, 60, 78] #updating
scores[0] = 90
print(scores)

seating = [["A1", "A2"], ["B1", "B2"]] #nested example
print(seating)
print(seating[1][0])   # element accessed at row 1, column 0

marks = [55, 70, 65]

marks.append(80)         # add to end
print("After append:", marks)

marks.insert(0, 40)      # insert 40 at index 0
print("After insert:", marks)

marks.remove(70)         # remove first occurrence of 70
print("After remove:", marks)

popped = marks.pop()     # remove and return last item
print("Popped value:", popped)
print("After pop:", marks)

marks.sort()              # sort ascending
print("After sort:", marks)

marks.reverse()           # reverse order
print("After reverse:", marks)

print("Length of list:", len(marks))
```
